```python
# Find the main section first
section = driver.find_element(By.CSS_SELECTOR, 'section#mwBQ')

# Then find all links within that section by TAG_NAME
country_elems = section.find_elements(By.TAG_NAME, 'a')

# Filter out the 'change' links
countries = [elem.text.strip() for elem in country_elems 
             if elem.text.strip() and elem.text.strip() not in ['change', 'change source']]
```

```python
print(countries[:20])
print(f"Total: {len(countries)}")
```

```python
#put the characters into a dataframe
df = pd.DataFrame(countries, columns = ["country"])
```
